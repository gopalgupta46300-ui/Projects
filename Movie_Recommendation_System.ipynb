```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/kaggle/input/datasets/organizations/tmdb/tmdb-movie-metadata/tmdb_5000_movies.csv')
df.head(5)
```

```python
df.columns
```

```python
df['tags'] = df['overview'] + df['genres']
df.head()
```

```python
new_df = df[['id','title','tags']]

print("------------------- columns ---------------------")
print(new_df.columns)

print("------------------- info ---------------------")
print(new_df.info())
```

```python
print(new_df.head())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features = 10000, stop_words = 'english')
cv
```

```python
vec = cv.fit_transform(df['tags'].values.astype('U')).toarray()
vec
```

```python
from sklearn.metrics.pairwise import cosine_similarity
sim = cosine_similarity(vec)
sim
```

```python
dist = sorted(list(enumerate(sim[0])), reverse = True, key= lambda vec : vec[1])
```

```python
dist
```

```python
len(dist)
```

```python
for i in dist[0:5]:
    print(new_df.iloc[i[0]].title)
```

```python
def recommend(movie):
    # get id of movie
    idx = new_df[new_df['title']== movie].index[0]

    dist = sorted(list(enumerate(sim[idx])), reverse = True, key= lambda vec : vec[1])

    for i in dist[0:5]:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Avatar')
```

```python
recommend('Godzilla 2000')
```

```python
recommend('The Fifth Element')
```

```python
recommend('Iron Man')
```
